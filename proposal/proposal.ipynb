# 3) EDA (Exploratory data analysis) questions

**Shared question:** When do later rides offset Divvy stations’ morning net bike outflow, and how do rider groups, routes, and weather relate to recovery? We analyze September 2026 station-days, with morning defined as 6–10 a.m. and recovery measured from 10 a.m.–8 p.m.

**Sean Qin — Station recovery:** Which stations repeatedly have morning net departures, and how often do later net arrivals offset them? I will compare recovery distributions and full-recovery rates across stations and plot cumulative hourly flows. Sensitivity checks will use morning deficits of at least five bikes and alternative time windows.

**Daniel Cai — Rider contributions:** At morning-deficit stations, do members and casual riders reinforce or offset each other’s imbalances? I will decompose morning and later net flows by membership within the same station-days, comparing bike counts and net flows per 100 group-specific trip endpoints. I will check whether patterns persist within bike types.

**Surya Venkataraman — Route contributions:** How much later inbound traffic comes from stations that received a station’s morning departures? I will match directed station pairs, compare reverse-route arrivals with arrivals from other stations, and examine consistency across days. I will test minimum route-volume thresholds. Reverse-route trips do not identify returning individuals.

**Anjeli Noémi (Mimi) — Weather and day type:** Within stations, how does recovery differ between rainy and dry days, across temperatures, and between weekdays and weekends? I will compare recovery within day types and morning-deficit groups, reporting day-level variability and sample sizes. I will check sensitivity to rain thresholds and treat Labor Day separately.

Each member will produce a main figure, a supporting table or figure, and a sensitivity check. Preparation is divided among Sean (station events), Daniel (rider categories), Surya (routes and duplicates), and Mimi (timestamps and weather), with shared validation and peer review.

# 4) Data

**Primary dataset:** [September 2026 Divvy trips](https://divvy-tripdata.s3.amazonaws.com/202609-divvy-tripdata.zip), containing 770,562 rides. Timestamps and station IDs support flows and routes; membership and bike type support rider comparisons. We will check duplicates and invalid durations and count departures and arrivals at their respective event times. Missing start/end station IDs affect 20.8%/22.1% of rides: flow analyses retain known endpoints; route analysis requires both IDs. We will report coverage and exclusions and check sensitivity to low-activity stations.

**One external dataset:** [Meteostat daily weather](https://dev.meteostat.net/data/timeseries/daily.html) for one documented Chicago station during September 2026. Temperature and precipitation will be joined by Chicago calendar date for Mimi’s comparisons. We will verify coverage, distinguish missing precipitation from zero, and document model-filled values. Rainy days initially mean positive precipitation. Day type and Labor Day are derived from dates; no capacity, income, or geographic datasets are required.

For positive morning deficits, **recovery = later net arrivals / morning net departures**. Full recovery means a ratio ≥1; negative values indicate further net outflow. Daniel’s group counts sum to total flows. Weather variability will be assessed across days, since stations share daily weather.

# 5) Stakeholder

Divvy operations and MCDC’s community partners can use recurring imbalances, rider and route contributions, and weather associations to prioritize stations for further monitoring. Trip flows cannot establish empty stations, unmet demand, or required truck movements. Results describe one month; weather comparisons show associations, not causal effects.
